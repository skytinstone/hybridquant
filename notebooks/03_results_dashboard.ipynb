# 03. 결과 대시보드 / Results dashboard

results/ 아래의 보고서(results/reports/*.md)와 그림(results/figures/<stage>/*.png)을 순서대로 보여 줍니다. 실험 중간에도 실행할 수 있고 없는 결과는 건너뜁니다.

Shows every stage report under results/reports and its figures inline. Safe to run while experiments are still running; missing stages are skipped.

```python
import glob, os, sys
from pathlib import Path
from IPython.display import Markdown, Image, display
ROOT = Path.cwd().resolve()
while not (ROOT / 'pyproject.toml').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
REPORTS = ROOT / 'results' / 'reports'
FIGS = ROOT / 'results' / 'figures'
ORDER = ['smoke', 'smoke_0.8b', 'grid_0.8b', 'superadd_0.8b', 'layer_sweep_0.8b', 'grid_4b_core',
         'r2_ablation_0.8b', 'r2_c4_0.8b', 'r2_niah_0.8b', 'r2_scale_2b', 'r2_gsm8k_4b', 'r2_niah_4b',
         'r2_scale_9b', 'r2_falcon_1.5b', 'all']
found = {p.stem: p for p in REPORTS.glob('*.md')} if REPORTS.exists() else {}
stages = [s for s in ORDER if s in found] + sorted(s for s in found if s not in ORDER)
print('reports found:', stages or 'none yet')
print('jsonl rows:')
for f in sorted((ROOT / 'results').glob('*.jsonl')):
    print(f'  {f.name:40s} {sum(1 for l in open(f) if l.strip()):6d}')
```

## 단계별 보고서와 그림 / Stage reports and figures

```python
def show_stage(stage):
    display(Markdown(f'# {stage}'))
    display(Markdown(found[stage].read_text(encoding='utf-8')))
    for png in sorted((FIGS / stage).glob('*.png')) if (FIGS / stage).exists() else []:
        display(Markdown(f'**{png.stem}**'))
        display(Image(filename=str(png), width=760))

for stage in stages:
    if stage != 'all':
        show_stage(stage)
if not stages:
    print('아직 보고서가 없습니다 / no reports yet')
```

## 종합 / Summary

```python
if 'all' in found:
    show_stage('all')
else:
    print('summary.md 가 아직 없습니다 (run_all 종료 후 생성) / summary not generated yet')
```
